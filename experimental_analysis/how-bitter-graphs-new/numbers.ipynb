```python
# ── LSTM vs architectures compute ratios (from parameters.toml fits) ──────
import re as _re, numpy as _np

def _parse_toml_val(section, key, text):
    pattern = rf'\[scaling_laws\.{section}\].*?{key}\s*=\s*([\d.e+\-]+)'
    m = _re.search(pattern, text, _re.DOTALL)
    return float(m.group(1)) if m else None

with open('parameters.toml') as _f:
    _toml = _f.read()

# Scaling laws: L - E = A * C^alpha
_A_mod  = _parse_toml_val('modern_transformer', 'A', _toml)
_al_mod = _parse_toml_val('modern_transformer', 'alpha', _toml)
_A_lstm = _parse_toml_val('lstm', 'A', _toml)
_al_lstm= _parse_toml_val('lstm', 'alpha', _toml)
_A_hist = _parse_toml_val('transformer_2017', 'A', _toml)
_al_hist= _parse_toml_val('transformer_2017', 'alpha', _toml)

def _C(target_le, A, alpha): return (target_le / A) ** (1.0 / alpha)

# Loss thresholds (L - E, with E = 1.9 irreducible loss)
_thresholds = [('3.377', 3.377), ('2.5', 2.5)]

output += '\n'
for _label, _tl in _thresholds:
    _C_mod  = _C(_tl, _A_mod,  _al_mod)
    _C_lstm = _C(_tl, _A_lstm, _al_lstm)
    _C_hist = _C(_tl, _A_hist, _al_hist)
    _mod_to_lstm  = _C_lstm / _C_mod    # how much more compute LSTM needs vs Modern
    _lstm_to_hist = _C_lstm / _C_hist   # how much more compute LSTM needs vs Historical
    _lstm_to_retro = _mod_to_lstm / _lstm_to_hist  # ratio of the two gaps
    output += f'{_mod_to_lstm:>12.2f}x  Modern to LSTM compute ratio at L-E={_label}\n'
    output += f'{_lstm_to_hist:>12.2f}x  LSTM to Historical compute ratio at L-E={_label}\n'
    output += f'{_lstm_to_retro:>12.4f}x  LSTM to retro (Modern→LSTM / LSTM→Historical) at L-E={_label}\n'
    output += '\n'
```

```python
with open("numbers.txt", "w") as f:
    f.write(output)
```
